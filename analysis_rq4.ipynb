# RQ4 — SE Tools and Practices Practitioners Want to Adopt

**RQ4:** *What are the software engineering tools and practices that the data science practitioners want to adopt?*

**Analyses:**
1. Planned-adoption ranking ("plan to use") for all 15 SE practices
2. Gap analysis: plan-to-use vs current adoption
3. Open-ended responses: future SE methods
4. Open-ended responses: dream tool (for thematic analysis)


```python
import warnings
import numpy as np
import pandas as pd
import sys, os
sys.path.insert(0, os.path.abspath('.'))
from analysis_utils import load_data, METHOD_COLS, CURRENT_TH, PLAN_TH, PAST_TH

warnings.filterwarnings('ignore')
df = load_data()
N  = len(df)
print(f"Dataset loaded: {N} respondents")
```

## 1 & 2. Planned Adoption Ranking and Gap Analysis

```python
rows = []
for c, label in METHOD_COLS.items():
    if c not in df.columns:
        continue
    curr_n   = int((df[c] == CURRENT_TH).sum())
    plan_n   = int((df[c] == PLAN_TH).sum())
    past_n   = int((df[c] == PAST_TH).sum())
    curr_pct = curr_n / N * 100
    plan_pct = plan_n / N * 100
    rows.append({
        'Practice':        label,
        'Current (%)':     round(curr_pct, 1),
        'Plan to use (%)': round(plan_pct, 1),
        'Plan to use (n)': plan_n,
        'Past use (n)':    past_n,
        'Gap (pp)':        round(plan_pct - curr_pct, 1),
    })

plan_df = pd.DataFrame(rows).sort_values('Plan to use (%)', ascending=False).reset_index(drop=True)
display(plan_df)
```

```python
print("Top 5 practices by planned adoption:")
display(plan_df.head(5)[['Practice', 'Current (%)', 'Plan to use (%)', 'Plan to use (n)']])
```

## 3. Open-Ended: Future SE Methods

```python
FUTURE_COL = 'future_se_methods_open_ended_response'
if FUTURE_COL in df.columns:
    future = df[FUTURE_COL].dropna()
    future = future[future.str.strip().str.len() > 3].reset_index(drop=True)
    print(f"Substantive responses: {len(future)}")
    for i, r in enumerate(future, 1):
        print(f"[{i:02d}] {r.strip()}")
    future_df = pd.DataFrame({'id': range(1, len(future) + 1), 'response': future.str.strip()})
    future_df.to_csv('future_se_methods.csv', index=False)
    print(f"\nSaved to future_se_methods.csv")
else:
    print("Column not found.")
```

## 4. Open-Ended: Dream Tool Responses

```python
DREAM_COL = 'dream_tool_description_open_ended_response'
if DREAM_COL in df.columns:
    dream = df[DREAM_COL].dropna()
    dream = dream[dream.str.strip().str.len() > 3].reset_index(drop=True)
    print(f"Responses passing length filter: {len(dream)}")
    print("Note: further manual filtering required to remove non-substantive responses.\n")
    for i, r in enumerate(dream, 1):
        print(f"[{i:02d}] {r.strip()}")
    dream_df = pd.DataFrame({'id': range(1, len(dream) + 1), 'response': dream.str.strip()})
    dream_df.to_csv('dream_tool_responses.csv', index=False)
    print(f"\nSaved to dream_tool_responses.csv")
else:
    print("Column not found.")
```
